# Training and Validation

```python
import os
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error
```

### Load Dataset

```python
train_set = pd.read_csv("../nasa_battery_dataset/processed/train_features.csv")
test_set = pd.read_csv("../nasa_battery_dataset/processed/test_features.csv")
```

### Drop columns that leak target information


```python
leak_columns = ['RUL', 'RUL_x']  # remove any column that gives away RUL
X_train = train_set.drop(columns=leak_columns)
y_train = train_set['RUL']

X_test = test_set.drop(columns=leak_columns)
y_test = test_set['RUL']
```

### Split training into train 


```python
X_train_final, X_val, y_train_final, y_val = train_test_split(
    X_train, y_train, test_size=0.2, random_state=42
)
```


### Scale features


```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_final)
X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(X_test)  # Keep for final evaluation
```

### Linear Regression

```python
from sklearn.linear_model import LinearRegression

lr_model = LinearRegression()
lr_model.fit(X_train_scaled, y_train_final)

# Evaluate on validation set
y_val_pred = lr_model.predict(X_val_scaled)
val_mse = mean_squared_error(y_val, y_val_pred)
val_mae = mean_absolute_error(y_val, y_val_pred)

print("Linear Regression Validation MSE:", val_mse)
print("Linear Regression Validation MAE:", val_mae)

```

```python
results = {}
results['LinearRegression'] = {'MSE': val_mse, 'MAE': val_mae}
```

### Random Forest Regression 

```python
from sklearn.ensemble import RandomForestRegressor

# Train Random Forest
rf_model = RandomForestRegressor(
    n_estimators=100,  # number of trees
    random_state=42,
    n_jobs=-1           # use all CPU cores
)
rf_model.fit(X_train_scaled, y_train_final)

# Evaluate on validation set
y_val_pred_rf = rf_model.predict(X_val_scaled)
val_mse_rf = mean_squared_error(y_val, y_val_pred_rf)
val_mae_rf = mean_absolute_error(y_val, y_val_pred_rf)

print("Random Forest Validation MSE:", val_mse_rf)
print("Random Forest Validation MAE:", val_mae_rf)
results['RandomForest'] = {'MSE': val_mse_rf, 'MAE': val_mae_rf}
```

### Gradient Boost Regression

```python
from sklearn.ensemble import GradientBoostingRegressor

# Train Gradient Boosting Regressor
gb_model = GradientBoostingRegressor(
    n_estimators=200,    # number of boosting stages
    learning_rate=0.1,   # step size
    max_depth=3,         # maximum depth of each tree
    random_state=42
)
gb_model.fit(X_train_scaled, y_train_final)

# Evaluate on validation set
y_val_pred_gb = gb_model.predict(X_val_scaled)
val_mse_gb = mean_squared_error(y_val, y_val_pred_gb)
val_mae_gb = mean_absolute_error(y_val, y_val_pred_gb)

print("Gradient Boosting Validation MSE:", val_mse_gb)
print("Gradient Boosting Validation MAE:", val_mae_gb)

results['GradientBoosting'] = {'MSE': val_mse_gb, 'MAE': val_mae_gb}
```

### XGB Regression

```python
from xgboost import XGBRegressor

# Train XGBoost Regressor
xgb_model = XGBRegressor(
    n_estimators=200,
    learning_rate=0.1,
    max_depth=3,
    random_state=42,
    n_jobs=-1
)
xgb_model.fit(X_train_scaled, y_train_final)

# Evaluate on validation set
y_val_pred_xgb = xgb_model.predict(X_val_scaled)
val_mse_xgb = mean_squared_error(y_val, y_val_pred_xgb)
val_mae_xgb = mean_absolute_error(y_val, y_val_pred_xgb)

print("XGBoost Validation MSE:", val_mse_xgb)
print("XGBoost Validation MAE:", val_mae_xgb)

# Add to results dictionary
results['XGBoost'] = {'MSE': val_mse_xgb, 'MAE': val_mae_xgb}
```

### LightGBM Regression

```python
from lightgbm import LGBMRegressor

# Train LightGBM Regressor
lgb_model = LGBMRegressor(
    n_estimators=200,
    learning_rate=0.1,
    max_depth=3,
    random_state=42,
    n_jobs=-1
)
lgb_model.fit(X_train_scaled, y_train_final)

# Evaluate on validation set
y_val_pred_lgb = lgb_model.predict(X_val_scaled)
val_mse_lgb = mean_squared_error(y_val, y_val_pred_lgb)
val_mae_lgb = mean_absolute_error(y_val, y_val_pred_lgb)

print("LightGBM Validation MSE:", val_mse_lgb)
print("LightGBM Validation MAE:", val_mae_lgb)

# Add to results dictionary
results['LightGBM'] = {'MSE': val_mse_lgb, 'MAE': val_mae_lgb}
```

### Final Validation Result Summary

```python
print("Validation Results Summary:\n")
for model_name, metrics in results.items():
    print(f"{model_name}:")
    print(f"  Validation MSE: {metrics['MSE']:.4f}")
    print(f"  Validation MAE: {metrics['MAE']:.4f}\n")
```
