# Visualization of Model Prediction


```python
import pandas as pd
import matplotlib.pyplot as plt
```

### Load Predictions


```python
results_df = pd.read_csv("../nasa_battery_dataset/processed/rul_predictions.csv")
y_test = results_df["Actual_RUL"]
y_test_pred = results_df["Predicted_RUL"]
```

```python
# 1. Actual vs Predicted RUL Scatter Plot
plt.figure(figsize=(8,6))
plt.scatter(y_test, y_test_pred, alpha=0.6, color='blue')
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', linewidth=2)
plt.xlabel('Actual RUL')
plt.ylabel('Predicted RUL')
plt.title('Random Forest: Actual vs Predicted RUL')
plt.grid(True)
plt.show()
```

```python
# 2. Residuals Histogram
residuals = y_test - y_test_pred
plt.figure(figsize=(8,6))
plt.hist(residuals, bins=50, color='purple', alpha=0.7)
plt.title("Residuals Distribution")
plt.xlabel("Error")
plt.ylabel("Count")
plt.grid(True)
plt.show()
```

```python
# 3. Residuals vs Actual RUL
plt.figure(figsize=(8,6))
plt.scatter(y_test, residuals, alpha=0.6, color='green')
plt.axhline(0, color='red', linestyle='--')
plt.xlabel('Actual RUL')
plt.ylabel('Residual (Actual - Predicted)')
plt.title('Residuals vs Actual RUL')
plt.grid(True)
plt.show()
```

# Visualization of Model Prediction

## 1. Actual vs Predicted RUL
This scatter plot compares the actual RUL values to the predicted RUL values from the Random Forest model. Points close to the red dashed line indicate accurate predictions.

## 2. Residuals Distribution
This histogram shows the distribution of residuals (errors). Most residuals are close to zero, indicating the model predicts RUL accurately for most samples.

## 3. Residuals vs Actual RUL
This plot visualizes how prediction errors vary across the range of actual RUL values. Ideally, residuals should be randomly scattered around zero, showing no bias.

Red line: No error (perfect prediction).

Green dots: Actual prediction errors for each sample.

Dots close to the red line mean small errors and good model performance.
